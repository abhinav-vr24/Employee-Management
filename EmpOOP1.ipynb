```python
class employee:
    next_emp_id = 1001      # Class variable

    def __init__(self, department, address):
        self.empid = employee.next_emp_id
        employee.next_emp_id += 1
        self.empname = self.get_empname()
        self.salary = self.get_salary()
        self.department = department
        self.address = address



    def get_empname(self):
        temp = input("Enter Employee Name:")
        if len(temp) <= 0:
            raise AttributeError("Employee Name can not be empty")
        else:
            return temp


    # Raising the Error
    def get_salary (self):
        temp = int(input("Enter the Salary:"))
        if temp <= 0:
            raise ValueError("Salary Can not be -ve")
        else:
            return temp

        


    def getdeptdetails(self):        #Association with Department
        return (self.department.show_department_details())
    
    def getadressdetails(self):     #Association with Address
        return (self.address.Show_Address_Details())



    def show_emp_details(self):
        return (
            "Employee Name:", self.empname,
            "Employee ID:", self.empid,
            "Employee Salary:", self.salary,
            "Employee Department Details:", self.getdeptdetails(),
            "Employee Address Details:", self.getadressdetails()
            )



# Child Class1
class manager(employee):

    def __init__(self, department, address):
        super().__init__(department, address)
        self.perks = self.get_perks()


    def get_perks(self):
        temp = int(input("Enter the perk:"))
        return temp

    def show_manager_total_salary(self):
        return ("Total_Salary:",(self.salary + self.perks))


# Child Class2
class salesman(employee):

    def __init__(self, department, address):
        super().__init__(department, address)
        self.comission = self.get_commission()


    def get_commission(self):
            temp = int(input("Enter the commission:"))
            return temp

    def show_salesman_total_salary(self):
        return ("Total_Salary:", (self.salary + self.commission))
    
```

```python
class department:
    def __init__(self):
        self.deptnam = self.get_deptnam()
        self.deptid = self.get_deptid()        
        self.show_department_details()


    def get_deptnam(self):
        temp = input("Enter Department Name: ").strip()
        if len(temp) == 0:
            raise AttributeError("Department name cannot be empty")
        return temp

    

    def get_deptid(self):
        temp = input("Enter Department ID: ").strip()
        if len(temp) == 0:
            raise AttributeError("Department ID cannot be empty")
        return temp



    def show_department_details(self):
        return ("Department Name:",self.deptnam, "Department ID:", self.deptid )
```

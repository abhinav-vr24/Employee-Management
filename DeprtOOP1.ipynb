```python
class department:
    def __init__(self):
        self.deptnam = self.get_deptnam()
        self.deptid = self.get_deptid()        
        self.show_department_details()


    def get_deptnam(self):
        temp = input("Enter Department Name:")
        return temp

    def get_deptid(self):
        temp = input("Enter Department ID:")
        return temp


    def show_department_details(self):
        return ("Department Name:",self.deptnam, "Department ID:", self.deptid )
```

```python
class address:
    def __init__(self):
        self.strt = self.get_strt()
        self.cty = self.get_cty()
        self.pcode = self.get_pcode()
        self.Show_Address_Details()

    def get_strt(self):
        temp = input("Enter Street Name:")
        return temp

    def get_cty(self):
        temp = input("Enter City Name:")
        return temp

    def get_pcode(self):
        temp = input("Enter Pincode:")
        return temp

    def Show_Address_Details(self):
        return ("Street:", self.strt, "City:", self.cty, "Pincode:", self.pcode)
```
